### 4.1 Use `apply` on a DataFrame [1 pt]
Use `apply` to transform the values in the `species` column of `penguins` to lowercase.

```python
# YOUR CODE HERE
penguins['species'] = penguins['species'].apply(lambda x: x.lower())
```

### 4.2 Write a custom function to transform data [1 pt]
Write a function `grams_to_ounces` that converts grams to ounces. For this calculation, use the conversion factor of 1 gram to 0.035 ounces.

```python
# YOUR CODE HERE
def grams_to_ounces(grams):
    return grams * 0.035
```

```python
# hidden tests are within this cell
```

### 4.3 `apply` a custom function to a DataFrame [1 pt]
Create a new column in `penguins`, `body_mass_ounces`, that is the result of passing the column `body_mass_g` through your custom function `grams_to_ounces`. You should use the `apply` method in pandas to do this. 

```python
# YOUR CODE HERE
penguins['body_mass_ounces'] = penguins['body_mass_g'].apply(lambda x: grams_to_ounces(x))
```
